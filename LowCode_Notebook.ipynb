The global energy landscape is currently undergoing an unprecedented and highly complex transition. At the same time, the market is experiencing major structural shifts, including a coming wave of liquefied natural gas production, plateauing global oil demand, slower coal plant retirements due to policy support, and persistent bottlenecks in power grid infrastructure that constrain renewable integration.

To maximise portfolio returns and mitigate risk, Lumina’s investment committee requires rapid cross-sector intelligence. Evaluating an investment in a US natural gas facility, for example, requires analysts to synthesise insights spanning multiple energy markets. Currently, Lumina’s analysts rely on manual review of dense and highly technical International Energy Agency reports covering Coal, Oil, Gas, Electricity, and Renewables. Preparing a cross-sector investment briefing typically requires two to three days of work, during which analysts review five to ten separate reports and spend roughly 65% of their time locating relevant passages rather than performing financial modelling or scenario analysis. The process relies heavily on manual search across thousands of pages of PDFs and cross-referencing fragments of information between reports.

This workflow creates measurable operational inefficiencies and analytical risk. Citation traceability is inconsistent, subtle correlations between reports may be overlooked, and valuable analyst time is diverted away from high-value modelling and strategic evaluation. In fast-moving market conditions or competitive bidding windows, the delay inherent in manual synthesis can mean that investment opportunities are missed before a complete briefing is produced.

To remain competitive, Lumina requires an AI-assisted system capable of reducing briefing preparation time by at least 90%, retrieving relevant report passages with at least 90% accuracy, and generating fully grounded responses with complete source attribution and minimal hallucination risk. By transitioning from manual document review to a Retrieval Augmented Generation system grounded exclusively in authoritative IEA reports, Lumina can enable analysts to ask complex natural language questions and receive instant, synthesised, and fully cited strategic briefings. This approach would eliminate days of manual document review, reduce the risk of human oversight, and allow analysts to focus on financial modelling and strategic decision-making rather than time-consuming information retrieval.

# **Question Answering using LLM**

## **Define the response function**

This helper function standardises interaction with a chat based Large Language Model using explicit `SystemMessage` and `HumanMessage` objects via the LangChain interface and a Groq hosted model.

It ensures:

* Clear separation between system level instructions and user queries
* Consistent model configuration with fixed generation parameters
* Reusability across prompts, experiments, and evaluation workflows
* A simplified abstraction for integrating LLM calls within the analysis pipeline

The prompts are converted into structured message objects, passed to the model in sequence, and the model is invoked to generate a response.

The function returns the raw model output content as a string.

```python
def generate_response(system_prompt='', user_prompt=''):

    # Instantiate Groq hosted model via LangChain wrapper
    llm = ChatGroq(
        # model='llama-3.3-70b-versatile',      # uncomment this line if this is the model you want to use
        # model='llama-3.1-8b-instant',         # uncomment this line if this is the model you want to use
        groq_api_key=GROQ_API_KEY,
        temperature=__________,                 # set the value of temperature, use 0 for deterministic and consistent outputs, recommended for this case
        max_tokens=__________,                  # set the value of max_tokens, a value of around 1500 is sufficient
        model_kwargs={"top_p": __________}      # set the value of top_p, use 0.7 for balanced token sampling
    )

    # Wrap the system-level instruction
    system_message = SystemMessage(content=system_prompt)

    # Wrap the user task input
    human_message = HumanMessage(content=user_prompt)

    # Invoke the model with ordered messages
    response = llm.invoke([system_message, human_message])

    # Return raw model output as string
    return response.content
```

## **Define the RAG response function**

This helper function implements the generation stage of a Retrieval Augmented Generation pipeline.

It first retrieves the most relevant document passages from a vector database using semantic similarity search. The user query is used to perform a search in the `vectorstore` vector store, returning the top `k` most similar document chunks along with their similarity scores. The retrieved passages are then concatenated to form the contextual input for the language model.

It ensures:

* Retrieval of relevant contextual passages using vector similarity search
* Grounding of model outputs in retrieved document content
* Controlled generation through fixed parameters such as temperature, `top_p`, and maximum token length
* A structured prompting approach separating system behaviour and user instructions

The retrieved context and user query are inserted into the user prompt, while the system prompt defines the expected behaviour of the model. These prompts are wrapped as `SystemMessage` and `HumanMessage` objects and passed to the model in sequence.

The function returns the raw model output as a string.
